```python
class mydata:

    def __init__(self, data):
        self.data = data

    def mean(self):
        return sum(self.data) / len(self.data)

    def std(self):
        avg = self.mean()
        squared_differences = [(value - avg) ** 2 for value in self.data]
        total_squared_differences = sum(squared_differences)
        variance = total_squared_differences / len(self.data)
        standard_deviation = variance ** 0.5
        return standard_deviation


x = mydata([1, 2, 7, 5])

print(x.mean())
print(x.std())
```

```python
import numpy as np

def lag(a, n):
    result = np.empty(len(a))  # making empty array that is same length as a
    result[:n] = np.nan  # put nan in first n spots
    result[n:] = a[:-n]  # fill rest of spots with the shifted values
    return result  # return lagged array

a = np.array( [1, 6, 8, 4] )

print(lag(a, 1))
print(lag(a, 4))
```

```python
def lagGenerator(n):
    return lambda a: lag(a, n)

lag2 = lagGenerator(2)

a = np.array([1, 6, 8, 4])

print(lag2(a))
```
